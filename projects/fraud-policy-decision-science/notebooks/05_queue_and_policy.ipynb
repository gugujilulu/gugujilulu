# What is the value of a complete adaptive rule?

Inspect case conservation, terminal liability and saturation paths. The separate sequential demonstration validates history-weighted estimation in its own state world.


```python
import json
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
cfg = json.loads((ROOT / "configs/base.json").read_text())
```

```python
from fraud_case import operations, sequential

daily, cases = operations.simulate(cfg, "delayed", cfg["seed"])
print(daily[["value", "liability", "recovered"]].sum())
print("Outstanding cases:", int(cases.outstanding.sum()))
print("Coverage changes:", int(daily.saturation.diff().fillna(0).ne(0).sum()))
validation, logs = sequential.run(cfg["seed"], repetitions=5, n=1000)
validation
```
